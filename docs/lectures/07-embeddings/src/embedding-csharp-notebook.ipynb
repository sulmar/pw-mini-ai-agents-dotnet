# Pierwszy wektor na przykładzie owoców

Co jest bardziej podobne do czerwonego jabłka: zielone jabłko czy wiśnia? Najpierw zagłosujcie, potem sprawdźcie wynik.

Wybierz kernel **.NET (C#)** i uruchamiaj komórki po kolei za pomocą **Shift + Enter**.

## 1. Zapisujemy cechy jako liczby

Za „tak” wpisujemy **1**, za „nie” wpisujemy **0**. Kolejność cech jest taka sama dla wszystkich owoców.

| Owoc | Czy to jabłko? | Czy jest czerwony? | Czy ma twardą pestkę jak wiśnia? |
|---|---:|---:|---:|
| Czerwone jabłko | 1 | 1 | 0 |
| Zielone jabłko | 1 | 0 | 0 |
| Wiśnia | 0 | 1 | 1 |

Trzy liczby z każdego wiersza tworzą **wektor**.

```
string Lista(int[] wektor) => "[" + string.Join(", ", wektor) + "]";

var czerwoneJablko = new[] { 1, 1, 0 };
var zielonejablko = new[] { 1, 0, 0 };
var wisnia = new[] { 0, 1, 1 };

Console.WriteLine($"Czerwone jabłko: {Lista(czerwoneJablko)}");
Console.WriteLine($"Zielone jabłko: {Lista(zielonejablko)}");
Console.WriteLine($"Wiśnia: {Lista(wisnia)}");
```

## 2. Sprawdzamy, które pozycje się różnią

Porównujemy pierwszą liczbę z pierwszą, drugą z drugą i trzecią z trzecią. Zapisujemy **0** przy zgodności i **1** przy różnicy. Spróbuj podać obie listy przed uruchomieniem komórki.

```
int[] Roznice(int[] a, int[] b) => a.Zip(b, (x, y) => x == y ? 0 : 1).ToArray();

var roznicejablek = Roznice(czerwoneJablko, zielonejablko);
var roznicezWisnia = Roznice(czerwoneJablko, wisnia);

Console.WriteLine($"Czerwone i zielone jabłko: {Lista(roznicejablek)}");
Console.WriteLine($"Czerwone jabłko i wiśnia: {Lista(roznicezWisnia)}");
```

## 3. Sumujemy różnice

Suma to **odległość Hamminga**, czyli liczba pozycji, na których wektory się różnią. W tym ćwiczeniu każda cecha liczy się tak samo.

```
Console.WriteLine($"Czerwone i zielone jabłko: {roznicejablek.Sum()}");
Console.WriteLine($"Czerwone jabłko i wiśnia: {roznicezWisnia.Sum()}");
```

Zielone jabłko różni się od czerwonego tylko kolorem. Wiśnia ma ten sam kolor, ale różni się na dwóch pozostałych pozycjach. Według naszych trzech cech **zielone jabłko jest bardziej podobne do czerwonego**.

**Pytanie do grupy:** Który owoc będzie bliżej czerwonego jabłka, jeśli porównamy tylko kolor?

W tym ćwiczeniu sami wybraliśmy cechy i zapisaliśmy ich wartości. Tak powstała ręcznie zbudowana reprezentacja wektorowa.

## 4. Zamieniamy zdania na wektory

Teraz każda pozycja odpowiada słowu. Ustalamy kolejność: **jabłko, czerwone, zielone**. Wpisujemy **1**, gdy słowo występuje w zdaniu, i **0**, gdy go brakuje.

Przewidź wektory trzech zdań. Czy zmiana szyku zmieni wektor?

```
var slownik = new[] { "jabłko", "czerwone", "zielone" };

int[] WektorZdania(string zdanie)
{
    var slowa = System.Text.RegularExpressions.Regex.Matches(zdanie.ToLowerInvariant(), @"\p{L}+")
        .Select(dopasowanie => dopasowanie.Value).ToArray();
    return slownik.Select(s => slowa.Contains(s) ? 1 : 0).ToArray();
}

var zdanieCzerwone = "Jabłko jest czerwone.";
var zdanieZielone = "Jabłko jest zielone";
var zdanieInnySzyk = "Czerwone jest jabłko";

foreach (var zdanie in new[] { zdanieCzerwone, zdanieZielone, zdanieInnySzyk })
    Console.WriteLine($"{zdanie} → {Lista(WektorZdania(zdanie))}");
```

### Porównujemy zdania

Używamy tej samej funkcji co przy owocach. W drugim zdaniu znika „czerwone” i pojawia się „zielone”. Ile pozycji się zmieni?

```
var wektorCzerwone = WektorZdania(zdanieCzerwone);

foreach (var zdanie in new[] { zdanieZielone, zdanieInnySzyk })
{
    var rozne = Roznice(wektorCzerwone, WektorZdania(zdanie));
    Console.WriteLine($"{zdanie} → {Lista(rozne)} → suma: {rozne.Sum()}");
}
```

## 5. Co się stanie po dodaniu „nie”?

Porównaj „Jabłko jest czerwone” z „Jabłko nie jest czerwone”. Przewidź, czy odległość wzrośnie, a potem uruchom komórkę.

```
var zdanieZNegacja = "Jabłko nie jest czerwone.";
var wektorNegacji = WektorZdania(zdanieZNegacja);

Console.WriteLine($"{zdanieCzerwone} → {Lista(wektorCzerwone)}");
Console.WriteLine($"{zdanieZNegacja} → {Lista(wektorNegacji)}");
Console.WriteLine($"Liczba różnic: {Roznice(wektorCzerwone, wektorNegacji).Sum()}");
```

**Pytanie do grupy:** Co zgubiła nasza reprezentacja?

„Nie” nie ma własnej pozycji w słowniku, więc oba zdania dają ten sam wektor. Dodanie „nie” pozwoliłoby wykryć jego obecność. Nadal można zapytać, do której części zdania odnosi się ta negacja.

## 6. Podobieństwo cosinusowe na przykładzie owoców

Wracamy do wektorów owoców. Sprawdzimy, jaki kąt tworzy wektor czerwonego jabłka z wektorem zielonego jabłka, a jaki z wektorem wiśni. Większy cosinus oznacza mniejszy kąt.

| Owoc | Wektor |
|---|---|
| Czerwone jabłko | (1, 1, 0) |
| Zielone jabłko | (1, 0, 0) |
| Wiśnia | (0, 1, 1) |

### Czerwone jabłko i zielone jabłko

1. **Iloczyn skalarny.** Mnożymy liczby na tych samych pozycjach i sumujemy wyniki.

   $$1\cdot1+1\cdot0+0\cdot0=1+0+0=1$$

2. **Długości.** Podnosimy współrzędne do kwadratu, sumujemy i wyciągamy pierwiastek.

   $$\sqrt{1^2+1^2+0^2}=\sqrt{2},\qquad \sqrt{1^2+0^2+0^2}=1$$

3. **Podobieństwo.** Dzielimy iloczyn skalarny przez iloczyn długości.

   $$\frac{1}{\sqrt{2}\cdot1}=\frac{1}{\sqrt{2}}\approx0{,}707$$

### Czerwone jabłko i wiśnia

1. **Iloczyn skalarny.**

   $$1\cdot0+1\cdot1+0\cdot1=0+1+0=1$$

2. **Długości.** Wektor czerwonego jabłka ma długość $\sqrt{2}$. Dla wiśni otrzymujemy

   $$\sqrt{0^2+1^2+1^2}=\sqrt{2}.$$

3. **Podobieństwo.**

   $$\frac{1}{\sqrt{2}\cdot\sqrt{2}}=\frac{1}{2}=0{,}500$$

Obie pary mają po jednej wspólnej cesze o wartości 1. Wektor wiśni jest dłuższy od wektora zielonego jabłka, więc w drugim porównaniu dzielimy przez większą liczbę. **Zielone jabłko uzyskuje wyższy wynik.**

### Sprawdzamy obliczenia w C#

Funkcja $\arccos$ pozwala odzyskać kąt z wartości cosinusa. C# zwraca go w radianach, więc przeliczamy wynik na stopnie.

```
double Dlugosc(int[] wektor) => Math.Sqrt(wektor.Sum(x => (double)x * x));

double PodobienstwoCosinusowe(int[] a, int[] b)
{
    if (a.Length != b.Length)
        throw new ArgumentException("Wektory muszą mieć tyle samo współrzędnych.");

    var iloczynDlugosci = Dlugosc(a) * Dlugosc(b);
    if (iloczynDlugosci == 0)
        return double.NaN; // Cosinus jest nieokreślony dla wektora zerowego.

    return a.Zip(b, (x, y) => (double)x * y).Sum() / iloczynDlugosci;
}

var owoce = new (string Nazwa, int[] Wektor)[]
{
    ("Czerwone jabłko", czerwoneJablko),
    ("Zielone jabłko", zielonejablko),
    ("Wiśnia", wisnia),
};

foreach (var (nazwa, owoc) in owoce.Skip(1))
{
    var iloczyny = czerwoneJablko.Zip(owoc, (a, b) => a * b).ToArray();
    var iloczynSkalarny = iloczyny.Sum();
    var dl1 = Dlugosc(czerwoneJablko);
    var dl2 = Dlugosc(owoc);
    var podobienstwo = PodobienstwoCosinusowe(czerwoneJablko, owoc);
    var kat = Math.Acos(Math.Clamp(podobienstwo, -1.0, 1.0)) * 180 / Math.PI;

    Console.WriteLine(nazwa);
    Console.WriteLine($"  1. Iloczyny współrzędnych: [{string.Join(", ", iloczyny)}] → suma: {iloczynSkalarny}");
    Console.WriteLine($"  2. Długości wektorów: {dl1:F3} i {dl2:F3}");
    Console.WriteLine($"  3. Podobieństwo cosinusowe: {podobienstwo:F3} → kąt: {kat:F0}°");
}
```

## 7. Porównujemy każdą parę owoców

Poprzednia tabela zawierała **cechy owoców**. Teraz na obu osiach umieszczamy **owoce**, a w komórkach zapisujemy podobieństwo cosinusowe. Tak powstaje macierz podobieństwa.

**Pytanie do grupy:** Jaki wynik otrzymamy dla zielonego jabłka i wiśni?

```
Console.Write($"{"",19}");
foreach (var owoc in owoce)
    Console.Write($"{owoc.Nazwa,19}");
Console.WriteLine();

foreach (var wiersz in owoce)
{
    Console.Write($"{wiersz.Nazwa,19}");
    foreach (var kolumna in owoce)
        Console.Write($"{PodobienstwoCosinusowe(wiersz.Wektor, kolumna.Wektor),19:F3}");
    Console.WriteLine();
}
```

Zielone jabłko i wiśnia mają wynik **0**, bo żadna pozycja nie zawiera jedynki w obu wektorach. Na przekątnej są jedynki, bo każdy z tych wektorów porównujemy tam z nim samym. Macierz jest symetryczna, bo kolejność porównania nie zmienia wyniku.

### Wizualizacja wektorów owoców w 3D

Wykres rysuje wektory owoców z początku notebooka w układzie współrzędnych. Biblioteka Plotly.NET.CSharp tworzy go bezpośrednio w C#, a Plotly.NET.Interactive wyświetla go pod komórką.

Pierwsza komórka pobiera pakiety NuGet. Wymagają one kernela na .NET 10.

```
#r "nuget: Plotly.NET.CSharp, 0.13.0"
#r "nuget: Plotly.NET.Interactive, 5.0.0"
```

Druga komórka korzysta z wektorów owoców z sekcji 1. Rysuje trzy odcinki od początku układu i dwa łuki w 3D za pomocą `Chart.Line3D`.

```
using System.Numerics;
using Plotly.NET;
using Wykres = Plotly.NET.CSharp.Chart;

{
    Vector3 Punkt(int[] v) => new(v[0], v[1], v[2]);

    var czerwone = Punkt(czerwoneJablko);
    var zielone = Punkt(zielonejablko);
    var wektorWisni = Punkt(wisnia);

    GenericChart Linia(Vector3[] punkty, string nazwa, string kolor, bool znaczniki) =>
        Wykres.Line3D<double, double, double, string>(
            x: punkty.Select(p => (double)p.X),
            y: punkty.Select(p => (double)p.Y),
            z: punkty.Select(p => (double)p.Z),
            Name: nazwa,
            ShowMarkers: znaczniki,
            LineColor: Color.fromString(kolor),
            LineWidth: 5);

    GenericChart Luk(Vector3 od, Vector3 doWektora, float promien, string kolor)
    {
        var u = Vector3.Normalize(od);
        var v = Vector3.Normalize(doWektora);
        var cosinus = Math.Clamp(Vector3.Dot(u, v), -1f, 1f);
        var kat = Math.Acos(cosinus);

        // Kierunek prostopadły do u w płaszczyźnie obu wektorów.
        var w = Vector3.Normalize(v - cosinus * u);

        var punktyLuku = Enumerable.Range(0, 41).Select(i =>
        {
            var t = kat * i / 40;
            return promien * ((float)Math.Cos(t) * u + (float)Math.Sin(t) * w);
        }).ToArray();

        return Linia(punktyLuku, $"{kat * 180 / Math.PI:F0}°", kolor, false);
    }

    var wykresOwocow = Wykres.Combine(new[]
    {
        Linia(new[] { Vector3.Zero, czerwone }, "Czerwone jabłko (1, 1, 0)", "#c62828", true),
        Linia(new[] { Vector3.Zero, zielone }, "Zielone jabłko (1, 0, 0)", "#16834b", true),
        Linia(new[] { Vector3.Zero, wektorWisni }, "Wiśnia (0, 1, 1)", "#7e22ce", true),
        Luk(czerwone, zielone, 0.55f, "#16834b"),
        Luk(czerwone, wektorWisni, 0.70f, "#7e22ce")
    });

    display(wykresOwocow);
}
```

Osie X, Y i Z odpowiadają kolejno cechom „to jabłko”, „jest czerwone” i „ma twardą pestkę”. W legendzie zobaczysz obliczone kąty: 45° dla zielonego łuku i 60° dla fioletowego. Obracając wykres, możesz pokazać, że każdy łuk leży w płaszczyźnie swojej pary wektorów.

## 8. Prawdziwe wektory z EmbeddingGemma (Ollama)

Wcześniej wektory układaliśmy ręcznie. Tutaj model `embeddinggemma` zamienia zdania na wektory. Notebook wysyła zdania do lokalnej Ollamy i dostaje wektory. Klucz API nie jest potrzebny.

Przed uruchomieniem komórki:
1. Zainstaluj Ollamę i pobierz model: `ollama pull embeddinggemma` (około 622 MB).
2. Jeśli Ollama nie działa w tle, uruchom `ollama serve` i zostaw terminal otwarty.

Komórka pokazuje treść zdania odniesienia, pierwsze pięć współrzędnych i ranking zdań według podobieństwa do zdania odniesienia.

```
using System.Net.Http;
using System.Text;
using System.Text.Json;

var zdaniaDemo = new[]
{
    "Chcę przerwać studia na jeden semestr, a potem wrócić.",
    "Chcę zrobić sobie czasową przerwę w nauce na uczelni.",
    "Chcę zrezygnować ze studiów na stałe."
};

var klientOllama = new HttpClient();
var cialo = JsonSerializer.Serialize(new { model = "embeddinggemma", input = zdaniaDemo });
var zawartosc = new StringContent(cialo, Encoding.UTF8, "application/json");

var odpowiedz = await klientOllama.PostAsync("http://localhost:11434/api/embed", zawartosc);
odpowiedz.EnsureSuccessStatusCode();

var json = JsonDocument.Parse(await odpowiedz.Content.ReadAsStringAsync());
var embeddingiDemo = json.RootElement.GetProperty("embeddings").Deserialize<double[][]>()!;

double CosinusDemo(double[] a, double[] b)
{
    var iloczyn = a.Zip(b, (x, y) => x * y).Sum();
    var dlugoscA = Math.Sqrt(a.Sum(x => x * x));
    var dlugoscB = Math.Sqrt(b.Sum(x => x * x));
    return iloczyn / (dlugoscA * dlugoscB);
}

Console.WriteLine($"Zdanie odniesienia: {zdaniaDemo[0]}");
Console.WriteLine($"Liczba współrzędnych: {embeddingiDemo[0].Length}");
Console.WriteLine("Pierwsze 5 wartości: " +
    string.Join(", ", embeddingiDemo[0].Take(5).Select(x => x.ToString("F3"))));

var ranking = zdaniaDemo
    .Select((zdanie, indeks) => new { Zdanie = zdanie, Indeks = indeks })
    .Skip(1)
    .Select(pozycja => new
    {
        pozycja.Zdanie,
        Wynik = CosinusDemo(embeddingiDemo[0], embeddingiDemo[pozycja.Indeks])
    })
    .OrderByDescending(pozycja => pozycja.Wynik);

Console.WriteLine("\nRanking podobieństwa:");
foreach (var pozycja in ranking)
{
    Console.WriteLine($"{pozycja.Wynik:F3} — {pozycja.Zdanie}");
}
```

## 9. Parafraza i wspólne słowa, tym razem z EmbeddingGemma

Porównujemy pierwsze zdanie z dwoma pozostałymi. Wektory tym razem pochodzą z modelu, a nie z ręcznie ustawionych cech.

**Pytanie do grupy:** Które zdanie jest bliżej według podobieństwa, a które wyraża podobny zamiar?

Najpierw prosty wariant ze słowami. Każda pozycja oznacza obecność słowa w słowniku **przerwę, studia, zawiesić, naukę**. Zdanie „Chcę zrobić sobie przerwę w nauce i potem wrócić na studia.” ma wektor `(1, 1, 0, 0)`, parafraza `(0, 0, 1, 1)`, a zdanie o przerwie świątecznej `(1, 1, 0, 0)`.

Zobacz, co się dzieje przy tym samym cosinusie: parafraza wypada najgorzej, bo nie dzieli żadnego słowa ze zdaniem odniesienia.

```
var odniesienieSlowa = new[] { 1, 1, 0, 0 };
var parafrazaSlowa = new[] { 0, 0, 1, 1 };
var przerwaSwiatecznaSlowa = new[] { 1, 1, 0, 0 };

double PodobienstwoSlow(int[] a, int[] b)
{
    var iloczyn = a.Zip(b, (x, y) => (double)x * y).Sum();
    return iloczyn / (Math.Sqrt(a.Sum(x => (double)x * x)) * Math.Sqrt(b.Sum(x => (double)x * x)));
}

Console.WriteLine($"Słowa, parafraza: {PodobienstwoSlow(odniesienieSlowa, parafrazaSlowa):F3}");
Console.WriteLine($"Słowa, przerwa świąteczna: {PodobienstwoSlow(odniesienieSlowa, przerwaSwiatecznaSlowa):F3}");
```

```
using System.Net.Http;
using System.Text;
using System.Text.Json;

var zdaniaParafrazy = new[]
{
    "Chcę zrobić sobie przerwę w nauce i potem wrócić na studia.",
    "Chcę na jakiś czas zawiesić naukę na uczelni.",
    "Studia zaczynam po wakacjach, a zimą mam przerwę świąteczną."
};

var klientParafrazy = new HttpClient();
var cialoParafrazy = JsonSerializer.Serialize(new { model = "embeddinggemma", input = zdaniaParafrazy });
var odpowiedzParafrazy = await klientParafrazy.PostAsync(
    "http://localhost:11434/api/embed",
    new StringContent(cialoParafrazy, Encoding.UTF8, "application/json"));
odpowiedzParafrazy.EnsureSuccessStatusCode();

var jsonParafrazy = JsonDocument.Parse(await odpowiedzParafrazy.Content.ReadAsStringAsync());
var wektoryParafrazy = jsonParafrazy.RootElement.GetProperty("embeddings").Deserialize<double[][]>()!;

double CosinusParafrazy(double[] a, double[] b)
{
    var iloczyn = a.Zip(b, (x, y) => x * y).Sum();
    return iloczyn / (Math.Sqrt(a.Sum(x => x * x)) * Math.Sqrt(b.Sum(x => x * x)));
}

Console.WriteLine($"Zdanie odniesienia: {zdaniaParafrazy[0]}");

var rankingParafrazy = zdaniaParafrazy
    .Select((zdanie, indeks) => new { Zdanie = zdanie, Indeks = indeks })
    .Skip(1)
    .Select(pozycja => new
    {
        pozycja.Zdanie,
        Wynik = CosinusParafrazy(wektoryParafrazy[0], wektoryParafrazy[pozycja.Indeks])
    })
    .OrderByDescending(pozycja => pozycja.Wynik);

Console.WriteLine("\nRanking podobieństwa:");
foreach (var pozycja in rankingParafrazy)
{
    Console.WriteLine($"{pozycja.Wynik:F3} — {pozycja.Zdanie}");
}
```

## 10. Mapa zdań (UMAP)

UMAP zmniejsza wektory z EmbeddingGemma do dwóch wymiarów, żeby zobaczyć, które zdania są do siebie podobne. Do tej mapy potrzebujemy co najmniej 16 zdań. Poniższa lista ma 17 zdań z czterech tematów: studia, owoce, pogoda i praca. Ich wektory pobieramy z Ollamy.

Pakiet UMAP dodajemy w osobnej komórce, a potem wyświetlamy mapę.

```
#r "nuget: UMAP, 1.0.62830"
```

```
using System.Net.Http;
using System.Text;
using System.Text.Json;

var zdaniaUmap = new[]
{
    "Chcę zrobić sobie przerwę w nauce i potem wrócić na studia.",
    "Chcę na jakiś czas zawiesić naukę na uczelni.",
    "Zamierzam zrezygnować ze studiów na stałe.",
    "W tym semestrze biorę urlop od uczelni.",
    "Egzaminy zdam w styczniu, zanim zacznę wakacje.",
    "Czerwone jabłko jest słodkie i chrupiące.",
    "Zielone jabłko smakuje kwaśno.",
    "Wiśnie dojrzewają późnym latem.",
    "Na śniadanie zjadłem banana i pomarańczę.",
    "Jutro będzie padać cały dzień.",
    "Słońce świeciło przez cały weekend.",
    "Zimą temperatura spada poniżej zera.",
    "Wiatr w nocy był bardzo silny.",
    "Pracuję zdalnie z domu od poniedziałku.",
    "Spotkanie z zespołem odbędzie się o dziesiątej.",
    "Projekt trzeba oddać do końca miesiąca.",
    "Szef zaakceptował nasz plan na przyszły kwartał."
};

var klientUmap = new HttpClient();
var cialoUmap = JsonSerializer.Serialize(new { model = "embeddinggemma", input = zdaniaUmap });
var odpowiedzUmap = await klientUmap.PostAsync(
    "http://localhost:11434/api/embed",
    new StringContent(cialoUmap, Encoding.UTF8, "application/json"));
odpowiedzUmap.EnsureSuccessStatusCode();

var jsonUmap = JsonDocument.Parse(await odpowiedzUmap.Content.ReadAsStringAsync());
var embeddingiUmap = jsonUmap.RootElement.GetProperty("embeddings").Deserialize<double[][]>()!;

Console.WriteLine($"Pobrano {embeddingiUmap.Length} wektorów po {embeddingiUmap[0].Length} współrzędnych.");
```

```
using System.Linq;
using WykresUmap = Plotly.NET.CSharp.Chart;

{
    if (zdaniaUmap.Length != embeddingiUmap.Length)
        throw new System.InvalidOperationException(
            "Po zmianie listy zdań uruchom ponownie komórkę z Ollamą.");

    if (embeddingiUmap.Length < 16)
        throw new System.InvalidOperationException(
            "Do tej demonstracji przygotuj co najmniej 16 zdań.");

    var wektory = embeddingiUmap
        .Select(wektor => wektor.Select(x => (float)x).ToArray())
        .ToArray();

    var umap = new UMAP.Umap(numberOfNeighbors: 5);
    int liczbaEpok = umap.InitializeFit(wektory);

    for (int epoka = 0; epoka < liczbaEpok; epoka++)
        umap.Step();

    var punkty = umap.GetEmbedding();

    var mapa = WykresUmap.Point<double, double, string>(
        x: punkty.Select(p => (double)p[0]).ToArray(),
        y: punkty.Select(p => (double)p[1]).ToArray(),
        MultiText: zdaniaUmap,
        Name: "Zdania");

    display(mapa);
}
```

### Jak czytać mapę UMAP?

## 11. Wyszukiwanie semantyczne w dokumentach uczelni

To krótkie demo pokazuje, jak embeddingi wyszukują dokumenty według znaczenia, a nie według wspólnych słów. Mamy trzy fikcyjne fragmenty dokumentacji uczelni. Ich embeddingi pobieramy z Ollamy (EmbeddingGemma) w jednym żądaniu.

Najpierw uruchamiamy komórkę z dokumentami. Potem komórkę z pytaniem, którą można zmieniać podczas wykładu.

```
using System.Net.Http;
using System.Text;
using System.Text.Json;

var dokumentyUczelni = new[]
{
    (
        Tytul: "Urlop dziekański",
        Tresc: "Aby czasowo przerwać naukę i wrócić w kolejnym semestrze, "
             + "złóż w dziekanacie wniosek o urlop dziekański."
    ),
    (
        Tytul: "Rezygnacja ze studiów",
        Tresc: "Aby definitywnie zakończyć studia, złóż w dziekanacie "
             + "oświadczenie o rezygnacji z kierunku."
    ),
    (
        Tytul: "Przerwa świąteczna",
        Tresc: "W czasie przerwy świątecznej uczelnia nie prowadzi zajęć. "
             + "Daty rozpoczęcia i zakończenia przerwy podaje kalendarz akademicki."
    )
};

var klientWyszukiwania = new HttpClient();

async Task<double[][]> PobierzWektoryWyszukiwania(string[] teksty)
{
    var dane = JsonSerializer.Serialize(new
    {
        model = "embeddinggemma",
        input = teksty
    });

    using var zawartosc = new StringContent(dane, Encoding.UTF8, "application/json");
    using var odpowiedz = await klientWyszukiwania.PostAsync(
        "http://localhost:11434/api/embed", zawartosc);

    odpowiedz.EnsureSuccessStatusCode();

    using var wynikJson = JsonDocument.Parse(await odpowiedz.Content.ReadAsStringAsync());
    return wynikJson.RootElement.GetProperty("embeddings").Deserialize<double[][]>()!;
}

var wektoryDokumentow = await PobierzWektoryWyszukiwania(
    dokumentyUczelni.Select(d => $"{d.Tytul}. {d.Tresc}").ToArray());

Console.WriteLine($"Przygotowano embeddingi {wektoryDokumentow.Length} dokumentów.");
```
